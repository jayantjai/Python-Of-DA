Number of times a text appears in the String --- We use count() to find it
* count() - it takes text as an argument and it is case-sensitive.

```
c = "Python is easy to learn. Many people love python"
print(c.count("Python"))
```

String Transformation Methods

1.) replace() -- to is used to improve data quality.
--Let say we have price in our dataset which contains "," in place of "." and we want to replace it. Then we can use replace() for it.

* replace() takes old value and new value as argument.

* replace() is used only in the String datatype.
* It helps to clean data show that we can perform operations on it easily.

```
price = "137,89"
print(price.replace(",","."))
```

Challenge Clean Phone No.

We have a messy Phone no., which we have to clean it and convert it into "integer" datatype

```
phone_No = "+49 (176) 123-4567"
a = phone_No.replace("+","").replace("(","").replace(")","").replace(" ","").replace("-","")
print(int(a))
```

Split Method

* It is also used to clean or separate the required values from data.
* It takes a separator, on basis of it we split our string.

* In the output we will get the list of all the splited values.
* It is used whenever we want only the specific part from our data values.

For example, lets take a date and seperate it into year, month and day.

```
date = "2026-09-21"
a = date.split("-")
print(a)
```
